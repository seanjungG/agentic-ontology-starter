### 4. [Phase 5 · Stage 3: 물리 그래프 구축] `STARTER_ONTOLOGY_GRAPH` 메타모델 · 실행 GQL 문장 & 바인딩 변수 · 전체 인스턴스 토폴로지 (`%%bqsql` + `H.run_gql`)

```python
%%bqsql df_paths
GRAPH `seanjung-poc.starter_ontology_dw.STARTER_ONTOLOGY_GRAPH`
MATCH p = (a)-[e]->(b)
RETURN TO_JSON(p) AS path
```

```python
gql_cust_rules = """GRAPH `seanjung-poc.starter_ontology_dw.STARTER_ONTOLOGY_GRAPH`
MATCH (c:CustomerContractNode)-[:USES_DEVICE]->(d:DeviceNode),
      (c)-[:SUBSCRIBES_PLAN]->(p:PlanNode)<-[ap:APPLIES_TO_PLAN]-(r:DomainRuleNode),
      (ch:DocChunkNode)
WHERE c.customer_id = 'CUST_DEMO_001' AND ch.chunk_id = ap.source_chunk_id
RETURN c.customer_id AS customer_id, d.device_name AS device_name, p.plan_name AS plan_name,
       r.rule_id AS rule_id, ap.source_chunk_id AS chunk_id, ch.content_sha256 AS sha256_anchor
ORDER BY r.rule_id"""
_ = H.run_gql(
    gql_cust_rules,
    variables={
        "GRAPH": ("seanjung-poc.starter_ontology_dw.STARTER_ONTOLOGY_GRAPH", "BigQuery 프로퍼티 그래프 식별자"),
        "c.customer_id": ("'CUST_DEMO_001'", "CustomerContractNode 고객 필터 변수"),
        "ch.chunk_id": ("ap.source_chunk_id", "DocChunkNode SHA-256 앵커 조인 키"),
    },
    title="3-Hop ISO GQL 실행 예시 (고객 약정 → 단말/요금제 ← 도메인 규칙 + 근거 청크)",
)
nodes, edges = H.paths_to_graph(H.pdf(df_paths)["path"])
H.md(f"**실측 추출된 물리 그래프 노드 `{len(nodes)}`개 · 엣지 `{len(edges)}`개**")
H.draw_graph(nodes, edges, "STARTER_ONTOLOGY_GRAPH 전체 인스턴스 토폴로지 (5계층 레이어 배치)", legend=H.LABEL_COLORS)
```

### Q&A
- **Q.** 정형 DW 테이블과 비정형 정책 문서는 어떤 단계를 거쳐 에이전트 답변으로 연결되나요?
  - **A.** 8단계 엔지니어링 흐름(1. 준비 → 2. 분석 → 3. 파싱 → 4. 통합 → 5. 구축 → 6. 에이전트 구축 → 7. 답변 → 8. 검증)을 통해 원시 CSV 6종과 마크다운 가이드 2종이 BigQuery 10개 테이블(`starter_ontology_dw`), 3-Layer LLM-Wiki(`W1~W3`), `STARTER_ONTOLOGY_GRAPH`(25 노드 · 29 엣지) Ready-to-Use 리소스로 통합되고, 리소스 커넥터(`OntologyResourceConnector`)와 도메인 비즈니스 도구 3종(`DomainMcpServer`) 및 2-Pass Gemini(`gemini-4-argon`, `Gate G2-W`) 에이전트가 이를 조회해 답변합니다.

### Data Analysis Key Findings
- BigQuery `starter_ontology_dw` 내 마스터 4종(22행), 지식/규칙 2종(9행), 엣지 브릿지 4종(31행) 총 10개 테이블(62행)과 비식별 고객 ID(`CUST_DEMO_001~006`, 위반 `0건`)가 실측 확인되었습니다.
